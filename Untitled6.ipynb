```python
import numpy as np


# REINFORCEMENT LEARNING USING MDP
# VALUE ITERATION + BELLMAN EQUATION


grid = [
    ['S', '.', '.', '.'],
    ['.', 'X', '.', '.'],
    ['.', '.', '.', '.'],
    ['X', '.', 'X', 'G']
]

rows = 4
cols = 4
```

```python
states = []

for i in range(rows):
    for j in range(cols):
        if grid[i][j] != 'X':
            states.append((i, j))

print("States:")
print(states)
```

```python
actions = {
    'UP': (-1, 0),
    'DOWN': (1, 0),
    'LEFT': (0, -1),
    'RIGHT': (0, 1)
}

print("\nActions:")
print(list(actions.keys()))
```

```python
def get_reward(next_state):

    r, c = next_state

    # Goal
    if grid[r][c] == 'G':
        return 10

    # Normal movement
    return -1
```

```python
def get_next_state(state, action):

    r, c = state
    dr, dc = actions[action]

    new_r = r + dr
    new_c = c + dc

    # Boundary checking
    if new_r < 0 or new_r >= rows:
        return state

    if new_c < 0 or new_c >= cols:
        return state

    # Obstacle checking
    if grid[new_r][new_c] == 'X':
        return state

    return (new_r, new_c)
```

```python
gamma = 0.9

# Initialize V(s) = 0
V = {state: 0 for state in states}

iterations = 20

for iteration in range(iterations):

    new_V = V.copy()

    for state in states:

        # Goal state
        if grid[state[0]][state[1]] == 'G':
            new_V[state] = 0
            continue

        values = []

        for action in actions:

            next_state = get_next_state(state, action)

            reward = get_reward(next_state)

            # Bellman Equation
            value = reward + gamma * V[next_state]

            values.append(value)

        # Bellman Optimality Equation
        new_V[state] = max(values)

    V = new_V
```

```python
print("\nState Values:")
print("----------------")

for i in range(rows):

    row_values = []

    for j in range(cols):

        if grid[i][j] == 'X':
            row_values.append(" X ")

        else:
            value = V[(i, j)]
            row_values.append(f"{value:5.2f}")

    print(row_values)
```

```python
policy = {}

for state in states:

    if grid[state[0]][state[1]] == 'G':
        policy[state] = 'GOAL'
        continue

    best_action = None
    best_value = float('-inf')

    for action in actions:

        next_state = get_next_state(state, action)

        reward = get_reward(next_state)

        value = reward + gamma * V[next_state]

        if value > best_value:

            best_value = value
            best_action = action

    policy[state] = best_action
```

```python
print("\nOptimal Policy:")
print("----------------")

for i in range(rows):

    row_policy = []

    for j in range(cols):

        if grid[i][j] == 'X':
            row_policy.append(" X ")

        else:
            action = policy[(i, j)]

            symbols = {
                'UP': '↑',
                'DOWN': '↓',
                'LEFT': '←',
                'RIGHT': '→',
                'GOAL': 'G'
            }

            row_policy.append(symbols[action])

    print(row_policy)
```
